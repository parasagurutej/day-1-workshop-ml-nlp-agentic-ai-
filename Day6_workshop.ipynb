```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
age=[10,20,30,40,50]
salary=[102458,207856,375085,409252,504654]
id=[101,205,308,407,501]
```

```python
plt.figure(figsize=(10,10))
```

```python
plt.plot(age,salary,marker="X",label="age",color="#9856")
plt.plot(age,id,marker="^",label="id",color="#2546")
plt.legend()
plt.title("age vs salary vs id")
plt.xlabel("age")
plt.ylabel("salary")
plt.savefig("age_vs_salary_vs_id.png")
plt.show()
plt.title("age vs salary vs id")
plt.savefig("age_vs_salary_vs_id.png")
plt.show()
```

```python
#bar graph
age=[10,20,30,40,50]
salary=[102458,207856,375085,409252,504654]
id=[101,205,308,407,501]

plt.bar(age,salary)
plt.legend()
plt.title("age vs salary vs id")
plt.xlabel("age")
plt.ylabel("salary")
plt.title("age vs salary vs id")
plt.savefig("age_vs_salary_vs_id.png")
plt.show()
```

```python
#histogram
age=[10,20,30,40,50]
salary=[102458,207856,375085,409252,504654]
id=[101,205,308,407,501]
plt.figure(figsize=(5,5))

plt.hist(age,bins=30)
plt.legend()
plt.title("age vs salary vs id")
plt.xlabel("age")
plt.ylabel("salary")
plt.title("age vs salary vs id")
plt.savefig("age_vs_salary_vs_id.png")
plt.show()
```
